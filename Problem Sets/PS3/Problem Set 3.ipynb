```python
"""
Question A:  Is it true that vitamin C has the ability to cure or prevent the common cold? Or is
    it just a myth? There's nothing like an in-depth experiment to get to the bottom
    of it all.
    
    N0: Vitamin C cannot prevent the common cold.
    NA: Vitamin C can help prevent the common cold.
    One-tailed

Question B:  It has been stated that individuals who are taller have a larger shoe size than
    those who are shorter.
    
    N0: The height of a person has no effect on shoe size.
    NA: Taller people larger shoe sizes than shorter people.
    One-tailed

Question C:  It has been stated that the higher level of education you obtain (Bachelor's, 
    Master's, etc), you will make a higher salary in industry.
    
    N0: Level of education has no effect on salary.
    NA: Higher level's of education correlates to a higher salary.
    Two-tailed

Question D:  It has been stated that individuals are often more susceptible to the flu in the 
    winter than other seasons.
    
    N0: The frequencey of the flu among people is not related to the season.
    NA: The flu is more common in the winter than other seasons.
    Two-tailed

Question E:  It has been stated that there is an impact on the IQ of a child when the mother 
    smokes while pregnant.
    
    N0: A mother smoking while pregnant has no impact on the child's IQ.
    NA: A child's IQ is lowered if the mother smokes while present.
    One-tailed.

Question F:  It has been stated that when it gets really hot outside, there are more ice cream 
    sales. 
    
    N0: The temperature outside does not effect ice cream sales.
    NA: The hotter the temperature outside, the more ice cream sales are made.
    One-tailed.

Question G:  It has been stated that when the tread on your tires gets worse, you'll get less 
    traction when it rains or snows. 
    
    N0: The state of tire tread has no impact on traction when it rains or snows.
    NA: Worse tire tread condition results in lower traction when it rains or snows.
    One-tailed.
"""
```

```python
import pandas as pd
import scipy.stats as sp
import numpy as np
import statsmodels.stats.multicomp as stats

data = pd.read_csv("Quality.csv")

print(data.groupby("Group")["QualityOfLifeScore"].describe())


"""
H0: The new medical treatment will have no effect on mental health.
HA: The new medical treatment will increase mental health.
    One-tailed.
"""

control = data.query("Group == 'Ctrl'")
treatment = data.query("Group == 'Treat1'")

num = control["QualityOfLifeScore"]
den = treatment["QualityOfLifeScore"]

dfn = num.size - 1
dfd = den.size - 1

f = np.var(num, ddof=1) / np.var(den, ddof=1)
p = 1 - sp.f.cdf(f, dfn, dfd)

print("Variability")
print("===========")
print(f"F-Value: {f:.6}")
print(f"P-Value: {p:.6}\n")

# Ttest

result_ttest = sp.ttest_ind(num, den, equal_var=True)

print("T-Test")
print("======")
print(f"P-Value: {result_ttest.pvalue:.6}")

#The P-Value of the variability test shows that amount of variance in the data of the groups is not due to chance because p=0.123 > alpha=0.05

#The significant level of the t-test shows there was no significant change in the data between groups, because p=0.077 > alpha=0.05, so the null hypothesis should be accepted.
#(Medical treatment increases mental health)
```

```python
data = pd.read_csv("Stickiness.csv")

print(data.groupby("Group")["Stickiness"].describe())

"""
N0: There is no difference in stickiness between different sites.
NA: Different sites have differing amounts of stickiness.
    Two-tailed.
"""

page1 = data.query("Group == 'Page1'")
page2 = data.query("Group == 'Page2'")
page3 = data.query("Group == 'Page3'")
page4 = data.query("Group == 'Page4'")

result_levene = sp.levene(
    page1["Stickiness"],
    page2["Stickiness"],
    page3["Stickiness"],
    page4["Stickiness"]
)

print("Variability")
print("===========")
print(f"P-Value: {result_levene.pvalue:.6}\n")

result_anova = sp.f_oneway(
    page1["Stickiness"],
    page2["Stickiness"],
    page3["Stickiness"],
    page4["Stickiness"]
)

print("Anova")
print("=====")
print(f"P-Value: {result_anova.pvalue:.6}")

#The P-Value of the variability shows that amount of variance in the data of the groups is not due to chance because p=0.576 > alpha=0.05

#The significant p-value of the ANOVA test shows that there was a significant change in the data between groups, because p=0.007 > alpha=0.05, so the null hypothesis should be rejected
#(Different sites have differing amounts of stickiness)

results_pairwise = stats.pairwise_tukeyhsd(
    endog=data["Stickiness"],
    groups=data["Group"],
    alpha=0.05
)

print(results_pairwise)

#The pairwise test shows that the null hypothesis should be accepted between page1-page4 and between page2-page4 because there is no significant difference between them
#however for the other page pairs the null hypothesis should be accepted
```
