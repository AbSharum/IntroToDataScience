```python
import numpy as np
import pandas as pd
import datetime

data = pd.read_csv('CreditCardCustomers1.csv')
data
```

```python
print("Number of rows: ",data.shape[0],"\n")
date = datetime.date.today()
currentYear = date.year
custAge = currentYear - data["DOB_Year"]
data.insert(3,"Customer Age",custAge)
data
```

```python
print(data['Attrition_Flag'].unique(),"\n")
print(data["Education_Level"].unique(),"\n")
print(data['Card_Category'].unique(),"\n")

print(data['Card_Category'].value_counts(),"\n")

print("Number of Silver card Members:",data[data['Card_Category'] == 'Silver'].shape[0],"\n")
print("Number of Customers born after 1990:",data[data['DOB_Year'] > 1990].shape[0],"\n")
print("Number of Customers born between 1980 & 1985:",data[data['DOB_Year'].between(1980,1985)].shape[0],"\n")

print(data['Income_Category'].unique(),"\n")
print("Customers Income greater or equal to $60k and less or equal to $120k:", data[data['Income_Category'].isin(['$60K - $80K','$80K - $120K','$120K +'])].shape[0],"\n")

```
