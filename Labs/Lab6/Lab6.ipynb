```python
import pandas as pd
import numpy as np
import datetime as datetime
import statsmodels.stats.multicomp as stats
import scipy.stats as sp
import matplotlib.pyplot as plt

#Import dataset
imdbData = pd.read_csv('imdb_23.csv')
salaryData = pd.read_csv('Salary_Data.csv')

imdbData.fillna(0,inplace=True)
```

```python
#Regression analysis on the Salary Data Set

regress = sp.linregress(salaryData['yearsofexperience'], salaryData['basesalary'])
#The linear regression equation is y = b_0 + b_1 * x_1
line = f'Regression line:y={regress.intercept:.2f}+ {regress.slope:.2f}x,r={regress.rvalue:.2f}'
line
```

```python
print("R-Value:\t", regress.rvalue)
print("P-Value:\t", regress.pvalue)
#The R-value is < 0.7 which means it's a weak model fit
#The P-value is < 0.05 which means it's a significant model
```
