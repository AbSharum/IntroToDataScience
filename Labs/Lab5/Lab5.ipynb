```python
import pandas as pd
import numpy as np
import datetime as datetime
import statsmodels.stats.multicomp as stats
import scipy.stats as sp


imdbData = pd.read_csv('imdb_23.csv')
salaryData = pd.read_csv('Salary_Data.csv')

imdbData.fillna(0,inplace=True)
salaryData.describe()
```

```python
imdbData.describe()
```

```python
result = sp.pearsonr(salaryData['yearsofexperience'],salaryData['yearlySalary'])
print("Correlation Coefficient:\t", result[0])
print("P-Value:\t\t\t",result[1])

# The P-value is greater than the alpha of 0.05 which means the relationship is insignificant
# There is a moderate positive correlation based on the correltaion coefficent
```

```python
result = sp.pearsonr(imdbData['IMDB_Rating'],imdbData['Gross'])
print("Correlation Coefficient:\t", result[0])
print("P-Value:\t\t\t",result[1])
# Our P-value is less than the alpha of 0.05 which means the relationship is significant
# There is a weak positive correlation based on our correlation coefficent
```
