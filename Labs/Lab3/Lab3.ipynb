```python
import pandas as pd
import numpy as np
import datetime 

readBuffer = "=" * 70

#Data Series

names = np.array(['Abraham','James','Carl','Mike','Wilson'])
stringNames = pd.Series(names)
print(stringNames , "\n")
print(readBuffer,"\n"+readBuffer)

#Date 

date = datetime.date.today()
print("Todays date is: ", date)
print("The month is: ", date.strftime('%B'))
print("The day of the week is: ", date.strftime('%A'), "\n")
print(readBuffer,"\n"+readBuffer)

#Data Frame

amazonBestSellOne = pd.read_csv('amazonBestSellers-File1.csv')
amazonBestSellTwo = pd.read_csv('amazonBestSellers-File2.csv')

print(amazonBestSellOne)
print(readBuffer,"\n"+readBuffer)
print(amazonBestSellTwo)
print(readBuffer,"\n"+readBuffer)

#Merging

amazonBestSell = pd.concat([amazonBestSellOne,amazonBestSellTwo])

print(amazonBestSell)
print(readBuffer,"\n"+readBuffer)

#Listing all Column Information

print(amazonBestSell.info())


```
