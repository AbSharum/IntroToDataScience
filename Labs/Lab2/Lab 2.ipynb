```python
instructor = {
    "CS1303": "B. Bright", 
    "GEOL3404": "M. Testa",
    "GEOL3504": "M. Testa",
    "CS1093": "B. Bright",
    "GEOL4023": "M. Testa"
}
room = {
    "CS1303": "BD143", 
    "GEOL3404": "MS204", 
    "GEOL3504": "MS106",
    "CS1093": "BD147",
    "GEOL4023": "MS116"
}
meetingTime = {
    "CS1303": "9:30-10:45AM", 
    "GEOL3404": "1:45 – 3:15PM",
    "GEOL3504": "4:15 – 5:25PM",
    "CS1093": "12:30 – 1:45PM",
    "GEOL4023": "ONLINE"
}


courses ="""
COURSES:
CS1303
GEOL3404
GEOL3504
CS1093
GEOL4023
==================
"""

print(courses,end = "")        
inpt = str(input("Enter course number:"))

for key in instructor:
    if(key.casefold() == inpt.casefold()):
        course = key
        roomNum = room[key]
        teacher = instructor[key]
        time = meetingTime[key]
        print("%-15s%5s\n%-15s%5s\n%-15s%5s\n%-15s%5s" 
      % ("Course:",course,"Room#:",roomNum,"Instructor:",teacher,"Time:",
         time))
        break
        
    else:
        print("Course does not exist")
        break
```
