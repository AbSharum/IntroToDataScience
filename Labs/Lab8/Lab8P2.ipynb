```python
import nltk
nltk.download()
```

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize

example = "This is my test. That I'm conducting. " \
            "More words to add. To the test. " \
            "Just a little more words"

#create sentences
sent_tokenize(example)
```

```python
#create words
word_tokenize(example)
```

```python
#filter words
from nltk.corpus import stopwords
tokenizedSentence = word_tokenize(example)
sw = set(stopwords.words("english"))
filter = []

for word in tokenizedSentence:
    if word.casefold() not in sw:
        filter.append(word)

filter
```

```python
#stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

stemmedWords = [stemmer.stem(word) for word in filter]

stemmedWords
```

```python
#lemmetizing
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatizedWords = [lemmatizer.lemmatize(word) for word in filter]

lemmatizedWords
```

```python
#speech tagging
nltk.pos_tag(tokenizedSentence)
```

```python
grammar = "NP:{<DT>?<JJ>*<NN>}"
parserChunking = nltk.RegexpParser(grammar)
taggedSentence = nltk.pos_tag(tokenizedSentence)
parserChunking.parse(taggedSentence)
output = parserChunking.parse(taggedSentence)
output.draw()
```
