```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
plt.rcParams.update({'font.size': 8})

data1 = pd.read_csv('CreditCardCustomers1.csv')
happiness = pd.read_csv('Happiness2019.csv')
happComb = pd.read_csv('HappinessCombined.csv')
```

```python
#Create new data frame using a query
data1_bar = data1.query('Education_Level == "High School" & Dependent_count > 0 & Marital_Status == "Single" & DOB_Year > 1980')

#Make a histogram with the new data frame
hist_data = data1_bar["Months_on_book"]
display = hist_data.plot.hist(color ='#002266')
display.set_title("Histogram", color ='#002266')
display.set_xlabel("Months on Book", color = 'red')
display.set_ylabel("Frequency", color ='red')
plt.show()
```

```python
#Create new data frame using a query
data1_bar2 = data1.query('DOB_Year > 1980 & Dependent_count > 3 & Marital_Status == "Married" ')
data1_bar2
#Make a histogram with the new data frame

bar_data = sns.countplot(data = data1_bar2, x = "Income_Category", hue ="Education_Level")
bar_data.set_xlabel("Income Category",color = 'red', fontsize = 10)
bar_data.set_ylabel("Frequency", color ='red', fontsize = 10)
bar_data.set_title("Count of Members by Income and Education",color ='#002266', fontsize = 10)
bar_data.legend(title = "Education Level", fontsize = 10)
```

```python
#Make new data frame using query 
data3_bar = happiness.query('Country == "Somalia" or Country == "Nicaragua" or Country == "Syria" or Country == "Finland"')
#Create bar graph
bar2 = plt.bar(data3_bar["Country"], data3_bar["Happiness Score"], color = ['blue','green','purple','yellow'])
plt.xlabel("Country",color='red', fontsize=10)
plt.ylabel("Happiness Score",color='red', fontsize=10)
plt.title('Happines Score by Country', color = "#002266", fontsize =10)
plt.show()
```

```python
#Make new data frame using query
data4 = happComb.query('Region == "Switzerland"')
#Create line graph

data4 = data4.sort_values(by=['DataYear'])

lineGraph = plt.plot(data4["LifeExpectancy"],data4["DataYear"],color='red',marker='D')
plt.yticks(data4["DataYear"])

plt.xlabel("Life Expectancy",color = 'red', fontsize=10)
plt.ylabel("Year",color="red",fontsize = 10)
plt.title('Life Expectancy of Switzerland', color="#002266",fontsize = 10)
```

```python
#Scatter plot with the same data set
scatter = plt.scatter(data4["LifeExpectancy"],data4["DataYear"],color='red',marker='H')
plt.yticks(data4["DataYear"])

plt.xlabel("Life Expectancy",color = 'red', fontsize=10)
plt.ylabel("Year",color="red",fontsize = 10)
plt.title('Life Expectancy of Switzerland', color="#002266",fontsize = 10)
```
