```python
class Workshops:
    days = 0
    regFee = 0
    name = ""

    def __init__(self,name,days,regFee):
        self.days = days
        self.regFee = regFee
        self.name = name

    def setDays(self,days):
        self.days = days

    def setRegFee(self,regFee):
        self.regFee = regFee

    def setName(self,name):
        self.name = name

    def getDays(self):
        return self.days
    
    def getRegFee(self):
        return self.regFee
    
    def getName(self):
        return self.name
    
class Location:
    lodgingFee = 0
    name = ""

    def __init__(self,name,lodgingFee):
        self.lodgingFee = lodgingFee
        self.name = name

    def setLodgingFee(self,lodgingFee):
        self.regFee = lodgingFee

    def setName(self,name):
        self.name = name

    def getName(self):
        return self.name

    def getLodgingFee(self):
        return self.lodgingFee

    

def main():

    #Setting all workshops with days and regestraion fees

    

    HandlingStress = Workshops("Handling Stress",3,1000)
    TimeManagment = Workshops("Time Managment",2,800)
    SuperVisionSkills = Workshops("Super Vision Skills",3,1200)
    Negotiation = Workshops("Negotiation",6,1500)
    InterviewSkills = Workshops("Interview Skills",1,300)

    shops = [HandlingStress,TimeManagment,SuperVisionSkills,Negotiation,InterviewSkills]

    #Setting all lodging fees

    Austin = Location("Austin",151)
    Chicago = Location("Chicago",225)
    Dallas = Location("Dallas",179)
    Orlando = Location("Orlando",322)
    Phoenix = Location("Phenix",175)
    Raleigh = Location("Raleigh",165)

    #Object arrays for ease of access

    shops = [HandlingStress,TimeManagment,SuperVisionSkills,Negotiation,InterviewSkills]
    locs = [Austin,Chicago,Dallas,Orlando,Phoenix,Raleigh]

    #Input

    print("%-25s%-25s%-25s\n" % ("Offered Workshops", "Number of Days", "Registration Fee"))


    for i in range(len(shops)):
        print("%-d- %-25s%-25d$%-25d" %(i+1,shops[i].getName(),shops[i].getDays(),shops[i].getRegFee()))

    print("\n%-25s%-25s\n" %("Locations","Lodging Fee"))

    
    for i in range(len(locs)):
        print("%-d- %-25s$%-25d" %(i+1,locs[i].getName(),locs[i].getLodgingFee()))

    print()


    ws = int(input("Enter chosen workshop (1-5): "))
    while (ws < 1 or ws >5):
        ws = int(input("Invalid input please enter again: "))

    loc = int(input("Enter chosen location (1-6): "))
    while (loc < 1 or loc >6):
        loc = int(input("Invalid input please enter again: "))

    for i in range(len(shops)):
        if (ws == (i+1)):
            sendShop = shops[i]

    for i in range(len(locs)):
        if (loc == (i+1)):
            sendLoc = locs[i]

#Calculate lodging price
            
    TotalLodge = sendLoc.getLodgingFee() * sendShop.getDays()
    Total = TotalLodge + sendShop.getRegFee()

#Output
    
    print("%-40s%6d\n%-40s$%6d\n%-40s$%6d /day\n%-40s$%6d\n%-40s\n%-40s$%6d" %("Active Days:",sendShop.getDays(),"Registration Fee:",sendShop.getRegFee(),"Lodging Fee",sendLoc.getLodgingFee(),"Total Lodging:",TotalLodge,"*"*55,"Total Cost:",Total))


    
main()
```
