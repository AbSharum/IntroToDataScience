```python
import pandas as pd
import numpy as np
import datetime as datetime

data = pd.read_csv('StuGrades.csv')
data
```

```python
data.info()
```

```python
data['dob'] = pd.to_datetime(data['dob'])
data.info()
```

```python
currentYear = datetime.datetime.today().year
data['age'] = currentYear - data['dob'].dt.year
data.info()
```

```python
data['avgScore'] = round((data['mathScore'] + data['readingScore'] + data['writingScore'])/3,2)
data
```

```python
data.describe()
```

```python
data.nlargest(5,'avgScore',keep='first')
```

```python
data.nsmallest(10,'avgScore',keep='first')
```

```python
query = data.query('age <= 17 & testPrepCourse == "none" & avgScore < 70')
query.info()
query
#I changed the query to check for age 17 instead of 16 because 16 returned 0. I beleive the pdf was made in 2023 so it being 2024 makes the people in the files a year older with respect
#to the datetime minus DOB calculation
```

```python
query.nsmallest(5,'avgScore',keep='first')
```
